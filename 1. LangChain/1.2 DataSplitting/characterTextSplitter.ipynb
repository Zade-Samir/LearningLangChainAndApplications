```python
from langchain_community.document_loaders import TextLoader

loader = TextLoader("receipt_record_1.txt")
docs = loader.load()
docs
```

```python
from langchain_text_splitters import CharacterTextSplitter

text_splitter = CharacterTextSplitter(
    separator = "\n",
    chunk_size = 100,
    chunk_overlap = 20
)

text = text_splitter.split_documents(docs)
text
```

```python
## manual way
speech = ""

with open("receipt_record_1.txt") as f:
    speech = f.read()

text_splitter = CharacterTextSplitter(
    separator = "\n", ## by default it is '\n\n'
    chunk_size = 100,
    chunk_overlap = 20
)

text = text_splitter.create_documents([speech])
text
```
