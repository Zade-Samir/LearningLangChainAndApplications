### Fetching the relevant texts/data which are required.

```python
from langchain_community.document_loaders import WebBaseLoader
import bs4

loader = WebBaseLoader(web_paths = ("https://www.langchain.com",),
                       bs_kwargs = dict(
                           parse_only = bs4.SoupStrainer(
                               class_ = ("t-heading-1-rg color-t-7fc8ff homepage")
                           )
                       )
                       )
```

```python
loader.load()
```

# Loading the Arxiv Files

```python
import arxiv
from langchain_core.documents import Document

client = arxiv.Client()

search = arxiv.Search(id_list=["1706.03762"])
results = list(client.results(search))
docs = []

if results:
    paper = results[0]

    doc = Document(
        page_content=paper.summary,
        metadata={
            "Title": paper.title,
            "Published": paper.published.strftime("%Y-%m-%d"),
            "Authors": ", ".join([a.name for a in paper.authors]),
            "Entry_ID": paper.entry_id,
        }
    )

    docs.append(doc)
    print(f"Successfully loaded paper: {doc.metadata['Title']}")
```
